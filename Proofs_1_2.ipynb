(a) The prox is firmly nonexpansive. 

**Proof:** 

Let (f) be a proper, lower semicontinuous, convex function.

Let us define
$p=\operatorname{prox}_{\tau f}(x), q=\operatorname{prox}_{\tau f}(y)$

By the optimality condition of the proximal operator, we have :
$$
p=\operatorname{pox}_{\tau f}(x) \iff 0\in\partial f(p) + \frac{p-x}{\tau}
$$
Therefore,
$$
\frac{x-p}{\tau}\in\partial f(p)
$$
Similarly,
$$
\frac{y-q}{\tau}\in\partial f(q)
$$
Since (f) is convex, its subdifferential (\partial f) is a monotone operator.
Hence,
$$\left\langle\frac{x-p}{\tau}-\frac{y-q}{\tau}, p-q\right\rangle\geq 0$$
Since $(\tau > 0)$, so we obtain 
$$
\langle x-p-y+q, p-q\rangle\geq 0
$$

$$
\langle x-y, p-q\rangle \geq -\langle -p+q, p-q\rangle
$$
$$
\langle x-y, p-q\rangle \geq \langle p-q, p-q\rangle \iff \langle p-q, x-y\rangle \geq \langle p-q, p-q\rangle
$$
Thus,
$$
\langle p-q, x-y\rangle\geq{||p-q||^2}
$$

$$
\langle \operatorname{prox}_{\tau f}(x) - \operatorname{prox}_{\tau f}(y), x-y \rangle\geq\left|\operatorname{prox}_{\tau f}(x)-\operatorname{prox}_{\tau f}(y)\right|^2
$$
Therefore, the prox is firmly nonexpansive.


(b) Prove that $x^*$ minimises $f+g$($f$ convex and differentiable) if and only if $ x^* = \operatorname{prox}_{\tau g}(x^* - \tau \nabla{f(x^*)})$ :

**Proof:**

We use the first-order optimality condition for minimizing the sum of a differentiable convex function and a convex function: 

$(\Rightarrow) :$
Let's suppose that $x^*$ minimizes $f+g$.
Then,
$$
0\in\nabla f(x^*) +\partial g(x^*)
$$
Equivalently,
$$
-\nabla f(x^*)\in\partial g(x^*)
$$
Since $\tau > 0$
$$
-\tau\nabla f(x^*)\in\tau\partial g(x^*)
$$
$$
\frac{(x^*-\tau\nabla f(x^*))-x^*}{\tau} \in\partial g(x^*)
$$
By the characterization of the proximal operator,
$$
p=\operatorname{prox}_{\tau g}(z) \iff \frac{z-p}{\tau}\in\partial g(p)
$$
By taking $ z = x^* - \tau\nabla f(x^*)$ and $p=x^*$, we obtain :
$$
x^*=\operatorname{prox}_{\tau g} \left(x^* - \tau\nabla f(x^*)\right)
$$

$(\Leftarrow) :$
Now let's suppose that $
x^* = \operatorname{prox}_{\tau g} \left(x^* - \tau\nabla f(x^*)\right)$

By the characterization of the proximal operator :
$$
\frac{(x^* - \tau\nabla f(x^*)) - x^*}{\tau} \in \partial g(x^*)
$$
$$
-\nabla f(x^*)\in\partial g(x^*)
$$
Therefore,
$$
0\in\nabla f(x^*) + \partial g(x^*)
$$
By the first-order optimality condition for convex functions,
$$
x^*\in\arg min_x(f(x) + g(x))
$$
